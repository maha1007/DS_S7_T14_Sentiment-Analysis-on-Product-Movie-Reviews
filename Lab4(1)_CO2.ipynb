# SESSION 2 - DATA INSPECTION


```python
# ==============================
import pandas as pd

df = pd.read_csv('your_data.csv')
print(df.head())
print(df.tail())
print(df.shape)
print(df.info())
print(df.isnull().sum())

# Iris dataset inspection
from sklearn.datasets import load_iris
iris_obj = load_iris()
df = pd.DataFrame(iris_obj.data, columns=iris_obj.feature_names)
df['species'] = iris_obj.target
df['species'] = df['species'].map({0:'setosa', 1:'versicolor', 2:'virginica'})
print(df.head())
print(df.info())
print(df.isnull().sum())

# Summary statistics
print(df.describe())
print(df.describe(include=['object']))
print(df['species'].value_counts())

# Data understanding
import matplotlib.pyplot as plt
sns.boxplot(x=df['sepal length (cm)'])
plt.show()
sns.pairplot(df, hue='species')
plt.show()
num_df = df.drop(columns='species')
sns.heatmap(num_df.corr(), annot=True, cmap='coolwarm')
plt.show()

# ==============================
```

# SESSION 5 & 6 - SEABORN


```python
# ==============================
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

iris = sns.load_dataset('iris')
sns.lineplot(x='sepal_length', y='sepal_width', data=iris)
plt.show()

# Styling + spines
sns.set_style('whitegrid')
sns.lineplot(x='sepal_length', y='sepal_width', data=iris)
sns.despine()
plt.show()

# Palette / countplot
tips = sns.load_dataset('tips')
sns.countplot(x='sex', data=tips, palette='Set1')
plt.show()

# Hue
sns.countplot(x='day', hue='sex', data=tips)
plt.show()

# Pairplots
sns.pairplot(tips, hue='day')
plt.show()
sns.pairplot(tips[['total_bill','tip','size']])
plt.show()

# KDE
x = np.random.randn(200)
sns.kdeplot(x=x, fill=True)
plt.show()
y = np.random.randn(200)
sns.kdeplot(x=x, y=y, fill=True)
plt.show()

# Distribution plots - modern replacements for deprecated distplot
sns.histplot(tips['total_bill'], bins=30, kde=False)
plt.show()
sns.displot(tips, x='total_bill', kde=True)
plt.show()
sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter')
plt.show()

# ==============================
```

# SESSION 7 - BOXPLOT & OUTLIERS


```python
# ==============================
import numpy as np
import matplotlib.pyplot as plt

data = np.array([10,12,15,18,20,22,25,30,100])
plt.boxplot(data)
plt.title('Box Plot')
plt.show()

# IQR method
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = data[(data < lower) | (data > upper)]
print('Q1:', Q1)
print('Q3:', Q3)
print('IQR:', IQR)
print('Lower bound:', lower)
print('Upper bound:', upper)
print('Outliers:', outliers)

# Filtered data (only after deciding an outlier should be excluded)
clean_data = data[(data >= lower) & (data <= upper)]
print(clean_data)

# ==============================
```

# SESSION 8 - HEATMAP & CORRELATION


```python
# ==============================
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = np.random.randint(1, 100, (10, 10))
sns.heatmap(data)
plt.show()

sns.heatmap(data, vmin=30, vmax=70)
plt.show()
sns.heatmap(data, cmap='coolwarm', center=50)
plt.show()
sns.heatmap(data, annot=True, fmt='d', cmap='Blues')
plt.show()

# Correlation methods
corr_df = tips[['total_bill','tip','size']]
print(corr_df.corr(method='pearson'))
print(corr_df.corr(method='spearman'))
print(corr_df.corr(method='kendall'))

# Correlation heatmap
numeric = iris.select_dtypes(include='number')
corr = numeric.corr(method='pearson')
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1, center=0)
plt.title('Correlation Heatmap')
plt.show()

# ==============================
```
