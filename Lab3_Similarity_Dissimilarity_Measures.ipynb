```python
#3. Similarity and Dissimilarity Measures for Data Analysis
#3.1.1.Vector-Based Distance Measures(Euclidean Distance)
import numpy as np
from scipy.spatial import distance

pointA=np.array([2,4,6])
pointB=np.array([5,1,9])

#Euclidean Distance - find how far these two points(pointA pointB) are from each other
euclidean_dist=distance.euclidean(pointA,pointB)
print("Euclidean Distance:",euclidean_dist)

#Similarity (inverse of distance)
similarity_euclidean=1/(1+euclidean_dist)
print("Euclidean Similarity:",similarity_euclidean)
```

```python
#3. Similarity and Dissimilarity Measures for Data Analysis
#3.1.2.Vector-Based Distance Measures(Manhatton Distance)
import numpy as np
from scipy.spatial import distance

pointA=np.array([2,4,6])
pointB=np.array([5,1,9])

#Manhatton Distance
manhatton_dist=distance.cityblock(pointA,pointB)
print("Manhatton Distance:",manhatton_dist)

#Similarity(inverse of distance)
similarity_manhatton=1/(1+manhatton_dist)
print("Manhatton Similarity:",similarity_manhatton)
```

```python
#3. Similarity and Dissimilarity Measures for Data Analysis
#3.1.3.Vector-Based Distance Measures(Minkowski Distance)
import numpy as np
from scipy.spatial import distance

pointA=np.array([2,4,6])
pointB=np.array([5,1,9])

#Minkowski Distance with p=3
minkowski_dist_p3=distance.minkowski(pointA,pointB,p=3)
print("Minkowski Distance (p=3):",minkowski_dist_p3)

#Similarity(inverse of distance)
similarity_minkowski=1/(1+minkowski_dist_p3)
print("Similarity Minkowski:",similarity_minkowski)
```
