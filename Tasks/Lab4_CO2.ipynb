```python
import matplotlib.pyplot as plt
y1=[] #stores +ve values
y2=[] #stores -ve values
x=range(-100,100,10) #Generates values from -100 to 90 with an interval of 10(-100,-90,-80..)
for i in x:y1.append(i**2) #y=x^2
for i in x:y2.append(-i**2) 
plt.plot(x,y1)
plt.plot(x,y2)
plt.xlabel("x")
plt.ylabel("y")
plt.ylim(-2000,2000) #yaxis limit
plt.axhline(0) #horizontal line
plt.axvline(0) #vertical line
plt.savefig("quad.png")
plt.show()
```

```python
#Dashboards in Python
import dash
#Dash Core components, used for interactive components such as graphs, dropdowns, sliders, buttons, etc.
from dash import dcc, html
#creates the new Dash application using variabl name called app
#__name__ is a special Python variable that identifies the current python module.
app=dash.Dash(__name__)
app.layout=html.Div([ #Layout defines what(structure/design of the webpage) appears on the dashboard.
    html.H1("Dash Tutorial"), #creates an HTML heading.
    dcc.Graph( #creates a graph.
        id="example",
        figure={
            "data":[
                {
                    "x":[1, 2, 3, 4, 5],
                    "y":[5, 4, 7, 4, 8],
                    "type": "line",
                    "name": "Trucks"
                },
                {
                    "x":[1, 2, 3, 4, 5],
                    "y":[6, 3, 5, 3, 7],
                    "type": "bar",
                    "name": "Ships"
                }
            ],
"layout": {
                "title": "Basic Dashboard"
            }
        }
    )
])
app.run(debug=True,port=8051)
                    
```

```python
import dash
from dash import dcc, html, Input, Output
app = dash.Dash(__name__)
app.layout = html.Div([
    html.Div("Square Calculator"),
    dcc.Input(
        id="input",
        type="text",
        placeholder="Enter a Number"
    ),
    html.Div(
        id="output"
    )
])
@app.callback(
    Output(component_id="output", component_property="children"),
    Input(component_id="input", component_property="value")
)
def update_value(input_data):
    try:
        return str(float(input_data) ** 2)
    except:
        return "Error: The input is not a number"
app.run(debug=False)
```
